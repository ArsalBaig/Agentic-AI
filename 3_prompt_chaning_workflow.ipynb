```python
import os
from langgraph.graph import StateGraph, START, END
from typing import TypedDict, Optional
from langchain_groq import ChatGroq
from dotenv import load_dotenv
```

```python
load_dotenv()
```

```python
model = ChatGroq(
    model= 'llama-3.3-70b-versatile',
    temperature= 0
)
```

```python
class BlogState(TypedDict, total= False):
    title : str
    outline : str
    content : str
```

```python
def create_outline(state: BlogState) ->BlogState:
    title = state['title'] 
    prompt = f'Generate a detailed outline for the following topic {title}'
    outline = model.invoke(prompt).content
    state['outline'] = outline
    return state
```

```python
def create_blog(state : BlogState) -> BlogState:
    title = state['title']
    outline = state['outline']

    prompt = f'Create a detailed blog on title {title} using the following outline {outline}'
    # Fixed: Changed .invoke() to .content
    content = model.invoke(prompt).content 
    state['content'] = content
    return state
```

```python
graph = StateGraph(BlogState)

graph.add_node('create_outline', create_outline)
graph.add_node('create_blog', create_blog)

graph.add_edge(START, 'create_outline')
graph.add_edge('create_outline', 'create_blog')
graph.add_edge('create_blog', END)

workflow = graph.compile()
```

```python
user_query = {'title' : 'Future of Ai'}
result = workflow.invoke(user_query)
print(result['outline'])
# print(result['content'])
```
